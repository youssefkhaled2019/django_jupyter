```python
from otm.models import Article,Reporter
```

```python
r = Reporter(first_name="John", last_name="Smith", email="john@example.com")
r.save()
r2 = Reporter(first_name="Paul", last_name="Jones", email="paul@example.com")
r2.save()
```

```python
from datetime import date
a = Article(id=None, headline="This is a test", pub_date=date(2005, 7, 27), reporter=r)
a.save()
```

```python
r = a.reporter
new_article = r.article_set.create(headline="John's second story", pub_date=date(2005, 7, 29))

print(new_article)
print(new_article.reporter)
print(new_article.reporter.id)
```

```python
new_article2 = Article.objects.create( headline="Paul's story", pub_date=date(2006, 1, 17), reporter=r )
print( new_article2.reporter)
print(new_article2.reporter.id)
print(r.article_set.all())
```

```python
r2.article_set.add(new_article2) # note add not create 
print(new_article2.reporter.id)
print( new_article2.reporter)
```

```python
print(r.article_set.all()) # r.article_set.last() ,list(r.article_set.all())[-1]
print(r2.article_set.all())
print(r.article_set.count())
print(r2.article_set.count())
```

```python
print( r.article_set.filter(headline__startswith="This") )

print( Article.objects.filter(reporter__first_name="John") )
print(  Article.objects.filter(reporter__first_name="John", reporter__last_name="Smith") )
```

```python
print(Article.objects.filter(reporter__pk=1))
print( Article.objects.filter(reporter=1) )
print( Article.objects.filter(reporter=r) )
print(  Article.objects.filter(reporter__in=[1, 2]) )
print( Article.objects.filter(reporter__in=[1, 2]).distinct() )
print( Article.objects.filter(reporter__in=[r, r2]).distinct() )
print(  Article.objects.filter(reporter__in=Reporter.objects.filter(first_name="John")).distinct() )
```

```python
print( Reporter.objects.filter(article__pk=1) )
print( Reporter.objects.filter(article=1) )
print( Reporter.objects.filter(article=a) )
print( Reporter.objects.filter(article__headline__startswith="This") )
print( Reporter.objects.filter(article__headline__startswith="This").distinct() )
print( Reporter.objects.filter(article__headline__startswith="This").count() )
print( Reporter.objects.filter(article__headline__startswith="This").distinct().count() )
print( Reporter.objects.filter(article__reporter__first_name__startswith="John") )
print( Reporter.objects.filter(article__reporter__first_name__startswith="John").distinct() )
print( Reporter.objects.filter(article__reporter=r).distinct() )
```
